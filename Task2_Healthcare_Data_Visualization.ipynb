```python
import pandas as pd

df = pd.read_csv("Synthetic_Medical_Dataset_v2 (1).csv")

print("Rows and columns:", df.shape)
```

```python
print(df.head())
```

```python
disease_counts = df["Predicted_Disease"].value_counts()

print(disease_counts)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=disease_counts.values,
    y=disease_counts.index
)

plt.title("Disease Distribution Among Patients")
plt.xlabel("Number of Patients")
plt.ylabel("Predicted Disease")

plt.tight_layout()
plt.show()
```

Observation: The disease distribution shows that Anxiety Disorder has the highest number of predicted cases (1,222), followed by Diabetes Type 2 and Urinary Tract Infection (1,195 each). Healthy individuals have the lowest count among the categories (1,084).

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.tight_layout()
plt.show()
```

Observation: The age distribution shows that patients are spread across different age groups. The graph forms a peak around the middle-age range, indicating that a larger number of patients belong to this age group.

```python
gender_counts = df["Gender"].value_counts()

print(gender_counts)
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Gender Distribution of Patients")
plt.show()
```

Observation: The gender distribution is nearly equal, with males accounting for 50.2% of the patients and females accounting for 49.8%. This shows a balanced representation of both genders in the dataset.

```python
# Step 9: Medication Usage Visualization

medicine_columns = ['Medicine_1', 'Medicine_2', 'Medicine_3']

medicine_counts = df[medicine_columns].melt(value_name='Medicine')['Medicine'].value_counts()

medicine_counts.head(10).plot(kind='bar', figsize=(10,5))

plt.title('Top 10 Most Frequently Used Medicines')
plt.xlabel('Medicine')
plt.ylabel('Number of Uses')
plt.xticks(rotation=45)
plt.show()
```

```python
# Step 10: Save the cleaned dataset

df.to_csv("Cleaned_Synthetic_Medical_Dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
