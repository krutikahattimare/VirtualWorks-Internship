```python
import pandas as pd

df = pd.read_csv("Synthetic_Medical_Dataset_v2.csv")

print("Rows and columns:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print(df.dtypes)
missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)

```

```python
# Fill missing medication fields with "Not Prescribed"
medication_cols = [
    'Medicine_1', 'Dosage_1', 'Frequency_1', 'Duration_1', 'Instructions_1',
    'Medicine_2', 'Dosage_2', 'Frequency_2', 'Duration_2', 'Instructions_2',
    'Medicine_3', 'Dosage_3', 'Frequency_3', 'Duration_3', 'Instructions_3'
]

df[medication_cols] = df[medication_cols].fillna("Not Prescribed")

print("Remaining missing values:", df.isnull().sum().sum())
```

```python
print("Total number of patients:", df['Patient_ID'].nunique())
```

```python
disease_counts = df['Predicted_Disease'].value_counts()

print("Most common diseases:")
print(disease_counts)
```

```python
average_age = df['Age'].mean()

print("Average patient age:", round(average_age, 2))
```

```python
df.to_csv("Cleaned_Synthetic_Medical_Dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
